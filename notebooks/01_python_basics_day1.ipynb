## 3. Calculations and operators

**Goal:** Calculate totals, averages, and remainders.

An **operator** is a symbol that performs an operation. An **expression**, such as `8 + 2`, produces a value.

| Operator | Operation | Example result |
|---|---|---|
| `+` | Addition | `8 + 2` → 10 |
| `-` | Subtraction | `8 - 2` → 6 |
| `*` | Multiplication | `8 * 2` → 16 |
| `/` | Division | `8 / 2` → 4.0 |
| `//` | Floor division | `9 // 2` → 4 |
| `%` | Remainder | `9 % 2` → 1 |
| `**` | Power | `3 ** 2` → 9 |

Multiplication precedes addition. Use parentheses to control grouping. Floor division rounds down: `-9 // 2` is -5. For our positive whole-number grouping examples, it counts complete groups. Do not divide by zero.

**Official reading:** [Numbers](https://docs.python.org/3/tutorial/introduction.html#numbers)


````python
score_1 = 60
score_2 = 75
score_3 = 90
total = score_1 + score_2 + score_3
average = total / 3
print("Total:", total)
print("Average:", average)
print("Without grouping:", 2 + 3 * 4)
print("With grouping:", (2 + 3) * 4)
````

## 4. Working with text

**Goal:** Combine text and present values in readable sentences.

Text is a **string**. `+` joins strings; `len()` counts their characters. A string method is a tool attached to a string: the dot in `name.upper()` selects the uppercase tool. `strip()` removes whitespace from both ends; these methods return new strings.

An **f-string** starts with `f` before the opening quote. Inside it, `{name}` inserts a value. `{score:.1f}` displays one digit after the decimal point. This formats the displayed text; it does not change the original score. Prefer f-strings when mixing text with numbers.

**Official reading:** [String methods](https://docs.python.org/3/library/stdtypes.html#string-methods) · [Formatted string literals](https://docs.python.org/3/tutorial/inputoutput.html#formatted-string-literals)


````python
raw_name = "  Ama  "
name = raw_name.strip()
score = 86.27
print("Hello, " + name)
print(name.upper())
print("Characters:", len(name))
print(f"{name} scored {score:.1f} marks.")
````

## 6. Comparisons and decisions

**Goal:** Run different instructions when a condition is true or false.

A **condition** is an expression evaluated to decide what to do. Numeric comparisons produce `True` or `False`.

| Symbol | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` / `<` | Greater than / less than |
| `>=` / `<=` | Greater than or equal / less than or equal |

`=` assigns; `==` compares. `if` tests a condition; `elif` tests another when earlier tests fail; `else` handles the remaining case. Only the first matching branch runs. Each header ends in `:`. Indent the instructions beneath it using four spaces. Indentation defines the block.

These score bands are invented for practice, not an official grading policy.

**Official reading:** [Comparisons](https://docs.python.org/3/reference/expressions.html#comparisons) · [if statements](https://docs.python.org/3/tutorial/controlflow.html#if-statements)


````python
score = 72
print("At least 50?", score >= 50)
if score >= 80:
    print("Excellent")
elif score >= 50:
    print("Pass")
else:
    print("Needs practice")
````

**Trace it:** 72 is below 80, so the first branch is skipped. It is at least 50, so the second branch prints Pass. No later branch runs.

## 7. Lists: several values in one place

**Goal:** Store, read, and update a collection of values.

A **list** holds an ordered collection. Write its items inside square brackets, separated by commas. An **index** identifies a position, starting at 0: `scores[0]` is the first score; `scores[-1]` is the last.

Lists can change. `scores.append(90)` adds an item at the end; `scores[1] = 80` replaces the second item. `len(scores)` counts items and `sum(scores)` adds numeric items. For a non-empty list, `sum(scores) / len(scores)` calculates the average. Avoid dividing by the length of an empty list.

**Official reading:** [Lists](https://docs.python.org/3/tutorial/introduction.html#lists) · [More on lists](https://docs.python.org/3/tutorial/datastructures.html#more-on-lists)


````python
scores = [60, 70, 80]
print("First:", scores[0])
scores.append(90)
scores[1] = 75
print("Updated:", scores)
print("Count:", len(scores))
print("Average:", sum(scores) / len(scores))
````

## 8. Loops: repeat a task

**Goal:** Apply the same instruction to several items.

A **loop** repeats a block of instructions. In a `for` loop, Python takes one item at a time from the collection and assigns it to the loop variable. The indented block runs once for each item.

`range(1, 4)` provides 1, 2, and 3: the stop value is excluded. `range(3)` provides 0, 1, and 2. Use a colon after the loop header and indent its body. Our examples use finite loops; while-loops can wait for conditions and are a later topic.

**Official reading:** [for statement](https://docs.python.org/3/reference/compound_stmts.html#the-for-statement) · [range()](https://docs.python.org/3/library/stdtypes.html#ranges)


````python
learners = ["Ama", "Kojo", "Esi"]
for learner in learners:
    print(f"Welcome, {learner}!")
print("All greetings complete.")
for number in range(1, 4):
    print(number)
````

**Trace it:** learner first becomes Ama, then Kojo, then Esi. The unindented completion message runs once after the greeting loop.

## 10. Optional: define a function

**Goal:** Make a reusable calculation.

A function lets you name a task and call it later. `def` introduces its definition. **Parameters** are the input names in the definition; **arguments** are the values supplied when you call it. `return` sends a result back to the calling code.

Defining a function does not run its body. Calling it does. `return` supplies a value you can store or reuse; `print()` displays a value. The returned value need not appear on screen unless you display it. Keep the function body indented.

**Official reading:** [Defining functions](https://docs.python.org/3/tutorial/controlflow.html#defining-functions)


````python
def average_two(first_score, second_score):
    return (first_score + second_score) / 2

result = average_two(60, 80)
print(result)
print(average_two(45, 55))
````

In this notebook, a numbered section is a lesson unit. In Python, a **module** groups code that can be imported and reused. The **standard library** is the collection of tools distributed with Python.

`import statistics` makes that module available under its name. The dot in `statistics.mean(scores)` selects its mean function, which calculates the arithmetic average. `statistics.median(scores)` finds the middle value, or the average of the two middle values for an even-sized collection. Use non-empty numeric data. This standard-library example needs no additional package installation.

**Official reading:** [Modules](https://docs.python.org/3/tutorial/modules.html) · [statistics](https://docs.python.org/3/library/statistics.html)


````python
import statistics

scores = [60, 70, 80]
print("Mean:", statistics.mean(scores))
print("Median:", statistics.median(scores))
````
